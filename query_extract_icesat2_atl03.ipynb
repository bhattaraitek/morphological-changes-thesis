```python
def extractAndCorrect(pth, local_datum: float = 0):
    '''
    Args:
    pth: path of h5 file
    local_datum: if required vertical datum is different from mean sea level, provided appropriate correction w.r.t. msl
    '''
    # def print_name(name):
    #     print(name)     
    with h5py.File(pth, 'r') as f:
        
        # f.visit(print_name)
    
        # List of the ground track names
        tracks = []
    
        for key in f.keys():             
        # Assuming ground tracks are top-level groups following a specific naming pattern
            if key.startswith('gt') and (key.endswith('l') or key.endswith('r')):                 
                height_path = f'{key}/heights/h_ph'
                if height_path in f:
                    tracks.append(key) 
                    
         # Dictionary to store the data for each track    
        tracks_data = {}
        if len(tracks) != 0:      
            # Iterate over each track name
            for track in tracks:
                # Construct the paths to the datasets for the current track
                height_path = f'{track}/heights/h_ph'
                latitude_path = f'{track}/heights/lat_ph'
                longitude_path = f'{track}/heights/lon_ph'
                quality_ph_path = f'{track}/heights/quality_ph'
                signal_conf_ph_path = f'{track}/heights/signal_conf_ph'
                geoid = f'{track}/geophys_corr/geoid'
                geoid_free2mean = f'{track}/geophys_corr/geoid_free2mean'
                dac = f'{track}/geophys_corr/dac'
                tide_ocean = f'{track}/geophys_corr/tide_ocean'
                tide_equilibrium = f'{track}/geophys_corr/tide_equilibrium'
                ph_id_count = f'{track}/heights/ph_id_count'
                dist_ph_along =f'{track}/heights/dist_ph_along'
                segment_ph_count = f'{track}/geolocation/segment_ph_cnt'
                segment_id = f'{track}/geolocation/segment_id'
                segment_dist_x = f'{track}/geolocation/segment_dist_x'
                ph_index_beg = f'{track}/geolocation/ph_index_beg'
                surf_type = f'{track}/geolocation/surf_type'
           
                # Read the data from each dataset
                heights = f[height_path][:]
                latitudes = f[latitude_path][:]
                longitudes = f[longitude_path][:]
                quality = f[quality_ph_path][:]
                signal_conf = f[signal_conf_ph_path][:]
                geoid = f[geoid][:],
                geoid_free2mean = f[geoid_free2mean][:],
                dac = f[dac][:],
                tide_ocean = f[tide_ocean][:],
                tide_equilibrium = f[tide_equilibrium][:]
                ph_id_count = f[ph_id_count][:]
                dist_ph_along =f[dist_ph_along][:]
                segment_ph_count = f[segment_ph_count][:]
                segment_id = f[segment_id][:]
                segment_dist_x =f[segment_dist_x][:]
                ph_index_beg = f[ph_index_beg][:]
                surf_type = f[surf_type][:]
    
                # heights_corrected = heights-(geoid+geoid_free2mean)-dac-tide_ocean-tide_equilibrium
                # print(np.array(ph_index_beg))
                
                #Extracting the segment id of photons. ph_idx_beg contains the starting index of photons for each segment
                ph_idx_beg_df = pd.DataFrame(ph_index_beg)
                ph_idx_beg_df['segment_id'] = ph_idx_beg_df.index
                ph_idx_beg_df.columns = ['ph_index_beg','segment_id']
                ph_idx_beg_df = ph_idx_beg_df.replace(0, pd.NA).dropna()
                
                heights_df = pd.DataFrame(heights)
                #matching the photon index with ph_idx_beg by adding 1 (not starting with 0)
                heights_df.index = heights_df.index + 1
                
                #This loop extracts index of photons lying on each segment. 
                # Then, it will add one columne with segment id, where each photon will have index of segment to which it belongs.
                # SInce, corrections like geoid are given segment wise, by identifying segment of each photon 
                # we can transfer corrections form segment to the photon and we can apply to each photon individually
                for i in range(ph_idx_beg_df.shape[0]-1):
                    idx = ph_idx_beg_df.iloc[i:i+2,:1].values.flatten()
                    seg_id = ph_idx_beg_df.iloc[i:i+1,1:2]
                    heights_df.loc[idx[0]:idx[1],'segment_id'] = seg_id.values[0][0]
                    heights_df.columns = ['heights','segment_id']
                    # print(seg_id.values[0][0])

                geoid_np = np.array(geoid).flatten()
                geoid_free2mean_np = np.array(geoid_free2mean).flatten()
                dac_np = np.array(dac).flatten()
                tide_ocean_np = np.array(tide_ocean).flatten()
                tide_equilibrium_np = np.array(tide_equilibrium).flatten()
                surf_type_np = surf_type
                
                geoid_series = pd.Series(geoid_np, index= np.arange(1, len(geoid_np)+1))
                geoid_free2mean_series = pd.Series(geoid_free2mean_np, index = np.arange(1,len(geoid_free2mean_np)+1))
                dac_series = pd.Series(dac_np,index = np.arange(1, len(dac_np)+1))
                tide_ocean_series = pd.Series(tide_ocean_np, index = np.arange(1, len(tide_ocean_np)+1))
                tide_equilibrium_series = pd.Series(tide_ocean_np, index = np.arange(1, len(tide_equilibrium_np)+1))
                surf_type_series =  pd.Series(list(surf_type_np), index = np.arange(1, len(surf_type_np)+1))
                #Mapping the correction values to photon based on segment id identified above
                heights_df['geoid'] = heights_df['segment_id'].map(geoid_series)
                heights_df['geoid_free2mean'] = heights_df['segment_id'].map(geoid_free2mean_series)
                heights_df['dac'] = heights_df['segment_id'].map(dac_series)
                heights_df['tide_ocean']= heights_df['segment_id'].map(tide_ocean_series)
                heights_df['tide_equilibrium'] = heights_df['segment_id'].map(tide_equilibrium_series)
                heights_df['surf_type'] = heights_df['segment_id'].map(surf_type_series)
                heights_df['heights_corrected'] = heights_df['heights'] - (heights_df['geoid']+heights_df['geoid_free2mean']
                                                                        )-local_datum#-heights_df['dac'] 
                # no data for these values, only inf -heights_df['tide_ocean']-heights_df['tide_equilibrium'], dac cprrection not relevant for land, only for ocean
        
                
                
                # Store the data in the dictionary
                tracks_data[track] = {

                    'latitudes': latitudes,
                    'longitudes': longitudes,
                    'heights': heights,
                    'ph_id_count':ph_id_count,
                    'dist_ph_along': dist_ph_along,
                    'segment_ph_cnt': segment_ph_count,
                    'segment_id': segment_id,
                    'segment_dist_x': segment_dist_x,
                    'ph_index_beg':ph_index_beg,
                    'heights_corrected': np.array(heights_df['heights_corrected']),
                    'quality_flg': quality,
                    'signal_conf': signal_conf,
                    'ph_geoid': np.array(heights_df['geoid']),
                    'ph_geoid_free2mean': np.array(heights_df['geoid_free2mean']),
                    'dac': np.array(heights_df['dac']),
                    'tide_ocean': np.array(heights_df['tide_ocean']),
                    'tide_equilibrium': np.array(heights_df['tide_equilibrium']),
                    'surf_type': np.array(heights_df['surf_type']),
                    'surf_type1': surf_type
                }

                
                
                # Optionally, print the first 5 values of each dataset for verification
                # print(f"{track} Heights:", heights[:5])
                # print(f"{track} Latitudes:", latitudes[:5])
                # print(f"{track} Longitudes:", longitudes[:5])
    return tracks_data
    
```

```python
def create_utm_grid(gdf: gpd.GeoDataFrame, resolution:float,current_coord_system='EPSG:4326', projected_system = 'EPSG:32645' , rotate_grid = True):
    
    '''
    Args:
    gdf: Geodataframe based on which grid will be created
    resolution: cell size of grid (square)
    current_coord_system: Coordinate system of gdf. If it is in spherical system (lat, lon), it needs to be reprojected so that we can calculate distance for grid creation
    projected_system: Project coordinate system like UTM
    rotate: If we want to roate grid so that it aligns in direction vector of gdf, we can turn it on.
    '''
    gdf.crs = current_coord_system
    gdf = gdf.to_crs(projected_system)  
    
    if rotate_grid:
        x = gdf.geometry.x
        y = gdf.geometry.y                    
        slope, intercept, r_value, p_value, std_err = linregress(x, y)

        # Create a line from the regression coefficients
        x_extent = np.array([x.min(), x.max()])
        y_extent = x_extent * slope + intercept
        # line = LineString(np.column_stack((x_extent, y_extent)))
        
        x_extent_vertical = np.array([x_extent.mean(),x_extent.mean()])
        y_extent_vertical = np.array([y_extent[0]- 5 * resolution,y_extent[1] + 5 * resolution])

        # line = LineString(np.column_stack((x_extent, y_extent)))
        line = LineString(np.column_stack((x_extent_vertical,y_extent_vertical)))
        
        angle_deg = np.degrees(np.arctan(slope))
        rotation = 90 + angle_deg if slope < 0 else angle_deg - 90
        xmin, ymin, xmax, ymax = line.bounds
        
        X = np.arange(xmin,xmax + resolution, resolution)
        Y = np.arange(ymin, ymax + resolution,resolution)
        
        grid_cells = [box(x,y, x + resolution, y + resolution) for x in X for y in Y]
        
        grid = gpd.GeoDataFrame(geometry= grid_cells)
        xmin, ymin, xmax, ymax = grid.total_bounds
        xc = (xmin + xmax)/2
        yc = (ymin + ymax)/2
        grid_rotated = grid.copy()
        grid_rotated['geometry'] = grid_rotated['geometry'].apply(lambda cell: rotate(cell, rotation, origin=(xc, yc)))
                
        # We have one issue with this rotation. The photon points do not passes through the centroid of each cell. So, 
        # during resampling the location of point is shifted from orignal one. Thus, it is necessary to translate grid
        # in such a way that ground track aligns with centroid of grid cells.   
        grid_centres = np.array([cell.centroid.coords[0] for cell in grid_rotated.geometry])
        # mid_x = x_extent.mean()
        # mid_y = y_extent.mean()
        mid_x = x_extent_vertical.mean()
        mid_y = y_extent_vertical.mean()

        dis = np.linalg.norm(grid_centres-np.array([mid_x, mid_y]), axis=1)
        min_distance_index = np.argmin(dis)
        nearest_grid_centre = grid_centres[min_distance_index]
        
        # calculating perpendicular distance from nearest grid centre to line
        per_dis = abs(slope * nearest_grid_centre[0] - nearest_grid_centre[1]+ (mid_y-slope*mid_x))/np.sqrt(slope**2 + 1)
        
        # To find quadrant of grid centre w.r.t line midpoint
        dx = mid_x- nearest_grid_centre[0]
        dy = mid_y - nearest_grid_centre[1]
        # slope2 = np.degrees(np.arctan2(dy, dx))
        sign_dx = dx/abs(dx)
        sign_dy = dy/abs(dy)
        
        theta = np.arctan(-1/slope)
        delta_x = sign_dx * abs(per_dis * np.cos(theta))
        delta_y  = sign_dy *abs(per_dis * np.sin(theta))
        grid_final = grid_rotated.copy()
        grid_final['geometry'] = grid_rotated['geometry'].translate(delta_x, delta_y) # now grid is aligned with ground track and points are aligned with centroid of the cells.
    else:  
        xmin, ymin, xmax, ymax = gdf.total_bounds
        x_coords = np.arange(xmin, xmax + resolution, resolution)
        y_coords = np.arange(ymin, ymax + resolution, resolution)
        grid_cells = [box(x, y, x + resolution, y + resolution) for x in x_coords for y in y_coords]
        grid_final = gpd.GeoDataFrame({'geometry': grid_cells}, crs=gdf.crs)
    return grid_final

def aggregate_points_to_grid(points_gdf, grid_gdf):
    # Spatial join points to grid
    joined_gdf = gpd.sjoin(points_gdf, grid_gdf, how="inner", op="within")
    # Aggregate the data
    aggregated_data = joined_gdf.groupby(joined_gdf.index_right).agg({
        'height_corrected': 'mean'  # or any other statistic
    })
    # Merge the aggregated data back with the grid
    final_grid = grid_gdf.join(aggregated_data, how="left")
    # final_grid['beam'] = points_gdf['beam']
    return final_grid
 
```

```python
def FilterN2Shp(tracks_data,resample,proj_crs,resolution):
    
    gdf_all_tracks = gpd.GeoDataFrame()
    # tracks = tracks_data.keys()
    if not tracks_data:
        print('\t---No photon heights available for this file---')
        return gpd.GeoDataFrame()
    else:
        for track in tracks_data.keys():
            print(f'\t ->Processing Ground track: {track}')
            geolocated_photons= []
            longitudes = tracks_data[track]['longitudes']
            latitudes = tracks_data[track]['latitudes']
            heights = tracks_data[track]['heights']
            heights_corrected = tracks_data[track]['heights_corrected']
            quality_flg = tracks_data[track]['quality_flg']
            signal_conf = tracks_data[track]['signal_conf']
            geoid = tracks_data[track]['ph_geoid']
            geoid_free2mean = tracks_data[track]['ph_geoid_free2mean']
            dac = tracks_data[track]['dac']
            tide_ocean = tracks_data[track]['tide_ocean']
            tide_equilibrium = tracks_data[track]['tide_equilibrium']
            surf_type = tracks_data[track]['surf_type']
            
            for lon, lat,hght,qf,sc,hght_corr,gd,gd_f2m,dac, td_oc,td_eq,stype in zip(longitudes,latitudes,heights, quality_flg,signal_conf,heights_corrected,geoid,geoid_free2mean,dac,tide_ocean,tide_equilibrium,surf_type):
                #Exporting only corrected photons to shapefile to reduce size of shape file
                
                point = {'geometry':Point(lon,lat),'height_corrected':hght_corr,'q_flg':qf,'sig_conf':sc,
                         'beam':track} 
                # if i want extra information i can add --,'height':hght,'geoid':gd,'geoid_free2mean':gd_f2m,'dac':dac,'tide_ocean': td_oc,'tide_equilibrium': td_eq, 'track': track,'q_flg':qf,'sig_conf':sc
                geolocated_photons.append(point)
                
            # Filtering out low qualituy photons. according to documentation of icesat2 0 represents nominal photons
            geolocated_photons_filtered = [photon for photon in geolocated_photons if 
                                           photon['q_flg'] == 0 
                                           and not math.isnan(photon['height_corrected']) 
                                           and not math.isinf(photon['height_corrected']) ]
                                        #    and ((photon['sig_conf'][0] in [4]) or (photon['sig_conf'][-1] in [4])) ]
            
            #selecting only high confidence signal from land 
            # geolocated_photons_filtered2 = [photon for photon in geolocated_photons_filtered if photon['sig_conf'][0] in [4]]  # if i want to filter inland water also i need to add this in array comprehension --or photon['sig_conf'][-1] in [4]
            
            gdf = gpd.GeoDataFrame(geolocated_photons_filtered)
            
            if not gdf.empty:
                gdfc = gdf.drop(columns=['sig_conf','q_flg'],axis=1)
                gdfc.crs = 'EPSG:4326'
                gdfc = gdfc.to_crs(proj_crs)
                
                if resample:
                    print(f'\t ---Resampling at {resolution}m---')
                    x = gdfc.geometry.x
                    y = gdfc.geometry.y                    
                    slope, intercept, r_value, p_value, std_err = linregress(x, y)

                    # Create a line from the regression coefficients
                    x_extent = np.array([x.min(), x.max()])
                    y_extent = x_extent * slope + intercept
                    # line = LineString(np.column_stack((x_extent, y_extent)))
                    
                    x_extent_vertical = np.array([x_extent.mean(),x_extent.mean()])
                    y_extent_vertical = np.array([y_extent[0]- 5 * resolution,y_extent[1] + 5 * resolution])

                    # line = LineString(np.column_stack((x_extent, y_extent)))
                    line = LineString(np.column_stack((x_extent_vertical,y_extent_vertical)))
                    
                    angle_deg = np.degrees(np.arctan(slope))
                    rotation = 90 + angle_deg if slope < 0 else angle_deg - 90
                    xmin, ymin, xmax, ymax = line.bounds
                    
                    X = np.arange(xmin,xmax + resolution, resolution)
                    Y = np.arange(ymin, ymax + resolution,resolution)
                    
                    grid_cells = [box(x,y, x + resolution, y + resolution) for x in X for y in Y]
                    
                    grid = gpd.GeoDataFrame(geometry= grid_cells)
                    xmin, ymin, xmax, ymax = grid.total_bounds
                    xc = (xmin + xmax)/2
                    yc = (ymin + ymax)/2
                    grid_rotated = grid.copy()
                    grid_rotated['geometry'] = grid_rotated['geometry'].apply(lambda cell: rotate(cell, rotation, origin=(xc, yc)))
                            
                    # We have one issue with this rotation. The photon points do not passes through the centroid of each cell. So, 
                    # during resampling the location of point is shifted from orignal one. Thus, it is necessary to translate grid
                    # in such a way that ground track aligns with centroid of grid cells.   
                    grid_centres = np.array([cell.centroid.coords[0] for cell in grid_rotated.geometry])
                    # mid_x = x_extent.mean()
                    # mid_y = y_extent.mean()
                    mid_x = x_extent_vertical.mean()
                    mid_y = y_extent_vertical.mean()

                    dis = np.linalg.norm(grid_centres-np.array([mid_x, mid_y]), axis=1)
                    min_distance_index = np.argmin(dis)
                    nearest_grid_centre = grid_centres[min_distance_index]
                    
                   # calculating perpendicular distance from nearest grid centre to line
                    per_dis = abs(slope * nearest_grid_centre[0] - nearest_grid_centre[1]+ (mid_y-slope*mid_x))/np.sqrt(slope**2 + 1)
                    
                   # To find quadrant of grid centre w.r.t line midpoint
                    dx = mid_x- nearest_grid_centre[0]
                    dy = mid_y - nearest_grid_centre[1]
                    # slope2 = np.degrees(np.arctan2(dy, dx))
                    sign_dx = dx/abs(dx)
                    sign_dy = dy/abs(dy)
                    
                    theta = np.arctan(-1/slope)
                    delta_x = sign_dx * abs(per_dis * np.cos(theta))
                    delta_y  = sign_dy *abs(per_dis * np.sin(theta))
                    grid_translated = grid_rotated.copy()
                    grid_translated['geometry'] = grid_rotated['geometry'].translate(delta_x, delta_y) # now grid is aligned with ground track and points are aligned with centroid of the cells.
                    # grid = create_utm_grid(gdf= gdfc, resolution= 30)
                    gdf_res = aggregate_points_to_grid(gdfc,grid_translated)
                    centroids = gdf_res.geometry.centroid
                    resampled_gdf_centroids = gpd.GeoDataFrame(gdf_res.drop(columns='geometry'),geometry=centroids)
                    resampled_gdf_centroids.crs = gdf_res.crs
                    gdf = resampled_gdf_centroids.dropna()
                    gdf.set_geometry = gdf['geometry']
                    
                else:
                    print('\t---Resampling turned off---')
                    gdf = gdfc
                print('\t ---Completed---.')
                gdf_all_tracks = gpd.GeoDataFrame(pd.concat([gdf_all_tracks,gdf], axis =0 , ignore_index= True))
            else:
                print('\t--- All photons flagged for this track. Low quality data. So, unable to generate shapefile---')        
        return gdf_all_tracks
    # gdf.iloc[:10000,:].to_excel('icesat2_data1.xlsx')
        
    
    
```

```python
root = short_name+'/'+dir_name_download
def find_h5_files(directory):
    h5_files = []
    # Walk through all directories and files in the specified directory
    for root, dirs, files in os.walk(directory):
        for file in files:
            # Check if the file ends with .h5
            if file.endswith('.h5'):
                # Join the root directory and file name to get the full path
                full_path = os.path.join(root, file)
                fn = os.path.splitext(os.path.basename(full_path))[0]
                splitted_fn = str.split(fn,'_')
                if "ATL03" in splitted_fn:
                    h5_files.append(full_path)
                # print(full_path)  # Optionally print each file path

    return h5_files

h5_files = find_h5_files(root)
h5_files
```
